```python
# Create a dataframe to store the predicted results
columns = ['dokument_aar', 'dokument_nr', 'embete', 'sidetall', 'index', 'type', 'height', 'width', 'x', 'y', 'ml_generated', 'ml_status']

# Create a DataFrame
df_store = pd.DataFrame(columns=columns)
```

```python
def format_output(df, folder_path, csv_file_path):
    ''' 
    Function to format the output of the model and store it in a CSV file.

    Args:
        df: DataFrame to store the output
        folder_path: Path to the folder containing the PDF files
        csv_file_path: Path to the CSV file to store the output
    '''

    # Loop through all the files in the folder and get predicted boxes
    for i, dokument in enumerate(os.listdir(folder_path)):
        
        # Get the name of the document
        pdf_path = folder_path + dokument

        # Remove .pdf from the name
        doc_id = dokument[:-4]

        print('Processing document nr.:', i, 'with id:', doc_id)

        # Split document id into year, number and embete
        elements = doc_id.split('_')
        year = int(elements[0])
        number = int(elements[1])
        embete = int(elements[2])

        # Get the predicted boxes
        images, predicted_boxes, texts = model_main.main(pdf_path)

        # Loop through all the predicted boxes
        for page_number, boxes in enumerate(predicted_boxes):
            for index, box in enumerate(boxes):
                # Get the box coordinates
                height, width, x, y = box

                # Define the data to be stored
                data = {
                    'dokument_aar': year,
                    'dokument_nr': number,
                    'embete': embete,
                    'sidetall': page_number,
                    'index': index,
                    'type': 'PERSONNUMMER',
                    'height': height,
                    'width': width,
                    'x': x,
                    'y': y,
                    'ml_generated': True,
                }

                # Add that if the box is empty it is not appended?
                if len(box) > 0:
                # Add the data to the DataFrame
                    df = pd.concat([df, pd.DataFrame([data])], ignore_index=True)

                print('Intermediate DataFrame:', df)

    # Write the DataFrame to a CSV file
    df.to_csv(csv_file_path, index=False)
```

```python
format_output(df_store, 'valideringssett/dokumenter/', 'predicted_output.csv')
```
